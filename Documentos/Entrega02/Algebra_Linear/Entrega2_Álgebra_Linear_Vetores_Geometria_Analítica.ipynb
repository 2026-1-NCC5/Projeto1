```python
import numpy as np
import pandas as pd
from google.colab import files

# 1. Escolher o arquivo CSV
print("Selecione o seu arquivo CSV original:")
uploaded = files.upload()

# Pega o nome do arquivo CSV
nome_do_arquivo = list(uploaded.keys())[0]

# 2. Carregar os dados do CSV para uma matriz NumPy
# header=None garante que ele não pule a primeira linha
matriz = pd.read_csv(nome_do_arquivo, header=None).values

# TRANSPOSTA (Diagonal)
# Troca linhas por colunas. O pixel (i, j) vai para (j, i)
matriz_diagonal = matriz.T

# ESPELHAMENTO LATERAL (Horizontal)
# Inverte apenas a ordem das colunas
matriz_espelhada_lat = np.fliplr(matriz)

# 3. Salvar os novos arquivos CSV para baixar
pd.DataFrame(matriz_diagonal).to_csv('matriz_diagonal_transposta.csv', index=False, header=False)
pd.DataFrame(matriz_espelhada_lat).to_csv('matriz_espelhada_lateral.csv', index=False, header=False)

print("\n✅ Processamento concluído!")
print("Arquivos prontos: diagonal, 180_graus e espelhada_lateral.")
```
